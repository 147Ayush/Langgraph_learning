```python
import random 
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, START, END
```

## Define State

```python
class State(TypedDict):
    graph_info: str
```

## Define the nodes

```python
# First node
def start_play(state: State):
    print("Start play node has been called.")
    return {
        "graph_info": state["graph_info"] + "I am planning to play."
    }

# Second node
def cricket(state: State):
    print("Cricket node has been called.")

    return{
        "graph_info" : state["graph_info"] + " I am planning to play cricket."
    }

# Third Node
def badminton(state : State):
    print("Badminton node has been called.")
    return{
        "graph_info" : state["graph_info"] + "I am planning to play badminton."
    }
```

## Define the conditional routing function

```python
def random_play(state : State) -> Literal["cricket", "badminton"]:
    if random.random() > 0.5:
        return "cricket"
    else:
        return "badminton"
```

## Create the StateGraph

```python
graph = StateGraph(State)

# adding nodes
graph.add_node("start_play", start_play)
graph.add_node("cricket", cricket)
graph.add_node("badminton", badminton)
```

## Add edges

```python
graph.add_edge(START, "start_play")
graph.add_conditional_edges("start_play", random_play)
graph.add_edge("cricket", END)
graph.add_edge("badminton", END)
```

## Compile

```python
app = graph.compile()
```

## Run

```python
result = app.invoke(
    {
        "graph_info": "My name is Ayush."
    }
)
print(result)
```

```python
# Visualize LangGraph
png_bytes = app.get_graph().draw_mermaid_png()

with open("langgraph_visualization.png", "wb") as f:
    f.write(png_bytes)

print("Graph visualization created successfully!")
```
