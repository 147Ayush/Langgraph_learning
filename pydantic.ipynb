## 5. Optional Fields

```python
from typing import Optional
```

```python
class Employee(BaseModel):
    id: int
    name: str
    department: str
    salary: Optional[float] = None
    is_active: bool = True
```

```python
employee = Employee(id = 202, name = "Ayush", department = "LLM Engineer")
print(employee)
```

## 7. List Fields

```python
class Classroom(BaseModel):
    room_number: str
    students: list[str]
    capacity: int
```

```python
classroom = Classroom(
    room_number="A1",
    students=["Alice", "Bob", "Charlie"],
    capacity=30
)
```

```python
print(classroom)
```

## 8. Handling ValidationError

```python
from pydantic import BaseModel, ValidationError

class Classroom(BaseModel):
    room_number : str
    students : list[str]
    capacity : int

try:
    classroom = Classroom(
        room_number="A1",
        students=[123],
        capacity=30
    )

except ValidationError as e:
    print(e)
```

## 9. Nested Pydantic Models ⭐

```python
class Address(BaseModel):
    street : str
    city : str
    zip_code : int

class Customer(BaseModel):
    name : str
    address : Address
```

```python
customer = Customer(name = "Ayush",
                    address = {
                        "street": "123 Main Street",
                        "city": "Bangalore",
                        "zip_code": "560001"
                    })
```

```python
print(customer)
```

## 10. Field() — Custom Validation Constraints ⭐

```python
from pydantic import BaseModel, Field
```

```python
class Student(BaseModel):
    name : str = Field(min_length=2, max_length=50) # string constraint
    marks : float = Field(gt = 0, lt = 101) # numeric constraint
```

```python
student = Student(name = "Ayush", marks = 100)
```

## 13. Default Values with Field

```python

class Student(BaseModel):
    name : str = Field(min_length=2, max_length=50) # string constraint
    marks : float = Field(gt = 0, lt = 101) # numeric constraint
    age: int = Field(
        default=18,
        description="User age"
    )
    email: str = Field(
        default="user@gmail.com",
        description="User email"
    )
```

```python
student = Student(name = "Ayush", marks = 100)
print(student)
```
